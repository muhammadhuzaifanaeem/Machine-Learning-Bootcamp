```python
import numpy as np

x = np.array([25, 34, 45, 29, 52, 38])
mean = x.mean()
std = x.std()
z = (x - mean) / std
print("Mean:", mean.round(3), "Std Dev:", std.round(3))
print("StandardScaler result:", z.round(3))
```

```python
x_min, x_max = x.min(), x.max()
x_scaled = (x - x_min) / (x_max - x_min)
print("Min:", x_min, "Max:", x_max)
print("MinMaxScaler result:", x_scaled.round(3))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder

numeric_features = ['age', 'income']
nominal_categorical = ['city']
ordinal_categorical = ['education']

numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

nominal_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

ordinal_transformer = Pipeline([
    ('ordinal', OrdinalEncoder(categories=[['Bachelors','Masters','PhD']]))
])

preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('nom', nominal_transformer, nominal_categorical),
    ('ord', ordinal_transformer, ordinal_categorical)
])
```
